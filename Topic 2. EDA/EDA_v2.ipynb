# Exploratory Data Analysis with a Scikit-Learn Dataset

This notebook shows a compact EDA workflow using the built-in Wine dataset from `sklearn.datasets`.

The goal is to inspect the dataset, summarize numeric features, check data quality, and visualize relationships before modeling.

## 1. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_wine

plt.style.use("ggplot")
pd.set_option("display.max_columns", 30)
```

## 2. Load the Dataset

```python
wine = load_wine(as_frame=True)

df = wine.frame.copy()
df["target_name"] = df["target"].map(dict(enumerate(wine.target_names)))

df.head()
```

## 3. Basic Structure

```python
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")

df.info()
```

```python
df.describe().T
```

## 4. Data Quality Checks

```python
missing_values = df.isna().sum().sort_values(ascending=False)
missing_values[missing_values > 0]
```

```python
duplicate_rows = df.duplicated().sum()
print(f"Duplicate rows: {duplicate_rows}")
```

## 5. Target Distribution

```python
class_counts = df["target_name"].value_counts().sort_index()
class_counts
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
class_counts.plot(kind="bar", ax=ax, color=["#4C78A8", "#F58518", "#54A24B"])
ax.set_title("Wine Classes")
ax.set_xlabel("Class")
ax.set_ylabel("Number of Samples")
ax.tick_params(axis="x", rotation=0)
plt.show()
```

## 6. Feature Distributions

```python
numeric_features = wine.feature_names

df[numeric_features].hist(figsize=(14, 10), bins=20, color="#4C78A8", edgecolor="white")
plt.suptitle("Numeric Feature Distributions", y=1.02, fontsize=16)
plt.tight_layout()
plt.show()
```

## 7. Compare Features by Class

```python
selected_features = ["alcohol", "malic_acid", "color_intensity", "proline"]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.ravel()

for ax, feature in zip(axes, selected_features):
    groups = [
        df.loc[df["target_name"] == class_name, feature]
        for class_name in wine.target_names
    ]
    ax.boxplot(groups, labels=wine.target_names, patch_artist=True)
    ax.set_title(feature.replace("_", " ").title())
    ax.set_xlabel("Class")
    ax.set_ylabel("Value")

plt.tight_layout()
plt.show()
```

## 8. Correlation Analysis

```python
corr = df[numeric_features].corr()

fig, ax = plt.subplots(figsize=(10, 8))
im = ax.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)

ax.set_xticks(np.arange(len(numeric_features)))
ax.set_yticks(np.arange(len(numeric_features)))
ax.set_xticklabels(numeric_features, rotation=90)
ax.set_yticklabels(numeric_features)
ax.set_title("Feature Correlation Matrix")

fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
plt.tight_layout()
plt.show()
```

## 9. Relationship Between Two Features

```python
fig, ax = plt.subplots(figsize=(8, 5))

for class_id, class_name in enumerate(wine.target_names):
    subset = df[df["target"] == class_id]
    ax.scatter(
        subset["alcohol"],
        subset["color_intensity"],
        label=class_name,
        alpha=0.8,
    )

ax.set_title("Alcohol vs. Color Intensity")
ax.set_xlabel("Alcohol")
ax.set_ylabel("Color Intensity")
ax.legend(title="Class")
plt.show()
```

## 10. Quick EDA Notes

- The dataset is small and fully numeric, which makes it convenient for introductory EDA.
- Class balance should be checked before training a classifier.
- Feature distributions and boxplots help reveal scale differences and possible outliers.
- Correlation analysis helps identify features that may carry similar information.
- Scatter plots can show whether two features separate target classes visually.
